# **Recursion**

A **recursive function** is a special type of function that calls itself to solve smaller instances of a problem until it reaches a **base case.**

**Purpose:** Solve problems that can be broken down into similar subproblems (like factorial, Fibonacci, tree traversal).

**Components of Recursion**

**Base Case:** The condition where the recursion stops. Without this, recursion goes on infinitely.

**Recursive Case:** The part where the function calls itself with a smaller or simpler input.


**Example: Print text for a given number of times**

```python
#iterative solution

def printntimes(str, n):
  while n > 0:
    print(str)
    n = n - 1

#drive code
printntimes("Hello, Iterative Print with loop", 5)
```

**Example: Factorial of a Number:**

```python
#iterative solution

def factorial(n):
  if n < 0:
    print("Wrong input")
    return
  elif n == 0:
    return 1
  else:
    fact = 1
    i = 1
    while i <= n:
      fact *=i # //fact=24*5=120
      i+=1
    return fact

f = factorial(5)
print(f)
```

```python
# Recursive solution

def factorial(n):
  if n < 0:
    print("Wrong input")
    return
  elif n == 0:
    return 1
  else:
    return n*factorial(n-1)


f = factorial(5)
print(f)
```

## **Iteration VS Recursion**

# **Time Requirement / Performance**

* In languages (e.g., C, Java, Python) that favor iterative looping constructs, there is usually significant time cost associated with recursive programs.
* As recursive calls have the overhead of managing the stack and making the function call (relative slowness of function calls).
* However, in functional languages, a function call is typically a very fast operation, and the difference is usually less noticeable.

# **Stack space Requirement**

*   Recursive algorithms tend to require more stack space than Iterative algorithms.
*   In some programming languages (e.g. **PYTHON**), the maximum size of the call stack is much less than the space available in the heap (ie, memory). A limit on the depth of recursion is placed to avoid **stack overflows**.

**Example: Power of a Number**

```python
#iterative solution

def power(n,p):
  if p == 0:
    return 1
  elif p > 0:
    pow = 1
    for i in range(p):
      pow *= n #pow = pow*n =3*3
    return pow
  else:
    pow = 1
    while p < 0 :
      pow *= 1/n #pow = 1/3 * 1/3
      p +=1
    return pow

power(3,2)
```

```python
# recursive solution
def power(n,p):
  if(p == 0):
    return 1
  elif p > 0:
    return n * power(n, p-1)
  else:
    return (1/n) * power(n, p+1)

power(3, 2)
```

**Example: Fibonacci Number (Two recursion branches):**

```python
# With Iteration

#0 1 1 2 3 5
def fibiter(n):

  # Check if input is 0 then it will print incorrect input
  if n < 0:
    print("Incorrect input")

  # Check if n is 0 then it will return 0
  elif n == 0:
    return 0

  # Check if n is 1 it will return 1
  elif n == 1:
    return 1
  else:
    f1=0
    f2=1
    tmp = int()
    for i in range(1,n):
        tmp = f1+f2
        f1=f2
        f2=tmp
    return tmp

fibiter(6)
```

```python
# With Recursion
# Function for nth Fibonacci number
def fib(n):

  # Check if input is 0 then it will
  # print incorrect input
  if n < 0:
    print("Incorrect input")

  # Check if n is 0
  # then it will return 0
  elif n == 0:       #Base part
    return 0

  # Check if n is 1
  # it will return 1
  elif n == 1 :       # Another base part
    return 1

  else:               # Recursive part
    return fib(n-1) + fib(n-2)

# Driver Program
print(fib(6))
```
