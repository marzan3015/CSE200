**An example to two dimentional list in python**

---



```python
# Creating a two-dimensional list (3x3 matrix)
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Accessing elements in the matrix
print(matrix[0][0])  # Prints the element at row 0, column 0 (1)
print(matrix[1][2])  # Prints the element at row 1, column 2 (6)

# Modifying elements in the matrix
matrix[1][1] = 10  # Changes the element at row 1, column 1 to 10

# Iterating through the matrix
for row in matrix:
    for element in row:
        print(element, end=' ')
    print()
```

```python
# Create an emply list
bookList = list()

#get book name, quantity and price for creating a book entry
aBook = ['The Last Journey', 10, 5000]

#add the book in the booklist
bookList.append(aBook)

# add three more books in the book list
aBook = ['Mind Your Own Business', 5, 15000]
bookList.append(aBook)

aBook = ['Travelling the World in 100 Days', 15, 25000]
bookList.append(aBook)

aBook = ['Get Me In the Team', 12, 10000]
bookList.append(aBook)

# print the booklist
print(bookList)

# print the name of all books (use loop)
for abook in bookList:
  print(abook[0])

#print the quantity and total price of all the books

for abook in bookList:
    print(abook[0])

# print the quantity and total price of all the books
print("\nQuantity and Total Price:")
for abook in bookList:
    total_price = abook[1] * abook[2]
    print("Book:", abook[0])
    print("Quantity:", abook[1])
    print("Total Price:", total_price)

# print unit price for the book 'Get Me In the Team'
print("\nUnit Price of 'Get Me In the Team':")
for abook in bookList:
    if abook[0] == 'Get Me In the Team':
        print(abook[2])

# print the book name with the longest title
longest_title = bookList[0][0]

for abook in bookList:
    if len(abook[0]) > len(longest_title):
        longest_title = abook[0]

print("\nBook with the longest title:")
print(longest_title)

# print all information for the book (name, quantity and price) having highest unit price
highest_price_book = bookList[0]

for abook in bookList:
    if abook[2] > highest_price_book[2]:
        highest_price_book = abook

print("\nBook with the highest unit price:")
print("Name:", highest_price_book[0])
print("Quantity:", highest_price_book[1])
print("Unit Price:", highest_price_book[2])
```

```python
# Creating a two-dimensional list (3x3 matrix)
matrix = [
    [1, 2, 3],
    [4, 5, 6,7],
    [7, 8, 9],
    [17, 8, 19, 27, 54],
]

# Print first value from list 1, first 2 values from list 2 and 3 values from list 3.. and continue this process to print a pyramid.
printcount = 1
for alist in matrix:
  i = 0
  while i < printcount:
      print(alist[i], end =', ')
      i +=1
  printcount +=1
  print()
```

**Using a two diemtional list store the name, id and CGPA of five students and then print each student data in python**

---



```python
# Creating a two-dimensional list to store student data
students = [
    ["John", 101, 3.8],
    ["Jane", 102, 3.7],
    ["Alice", 103, 3.9],
    ["Bob", 104, 3.5],
    ["Eva", 105, 3.6]
]

# Printing each student's data
for student in students:
    name = student[0]
    student_id = student[1]
    cgpa = student[2]
    print(f"Name: {name}, ID: {student_id}, CGPA: {cgpa}")

# Output:
# Name: John, ID: 101, CGPA: 3.8
# Name: Jane, ID: 102, CGPA: 3.7
# Name: Alice, ID: 103, CGPA: 3.9
# Name: Bob, ID: 104, CGPA: 3.5
# Name: Eva, ID: 105, CGPA: 3.6
```
