## **Dictionary in Python**

**▪ Dictionaries are used to store data values in key:value pairs.**

**▪ A dictionary is a collection which is ordered*, changeable and do not allow duplicates.**

**▪ Dictionaries are written with curly brackets, and have keys and values.**

```python
#the car dictionary
carDict =	{
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964,
  'color': ['red', 'green', 'white']
}

print(carDict)
```

**Duplicates Not Allowed**

```python
#the car dictionary
carDict =	{
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964,
  'color': ['red', 'green', 'white'],
  'year': [2000, 2007]                 ## its a duplicate entry as key values are the same
}

print(carDict)
```

###**Accessing Items in the Dictionary**

```python
carDict =	{
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964,
  'color': ['red', 'green', 'white']
}
```

```python
# using 'key' to access value
carbrand = carDict['brand']
print(carbrand)

carColor = carDict['color']
print(carColor)
```

```python
#using get() function
carbrand = carDict.get('brand')
print(carbrand)

carColor = carDict.get('color')
print(carColor)
```

**Get all the keys in a dictionary**

```python
#using keys() function
keys = carDict.keys()
print(keys)

#convert to a list
keyList = list(keys)
print(keyList)
```

**Get all the values in a dictionary**

```python
#using values() function
values = carDict.values()
print(values)

#convert to a list
valueList = list(values)
print(valueList)
```

**Loop over a dictionary**

-----------------------------------------------------

```python
# loop over a dictionary returs all the keys one-by-one
for x in carDict:
  print(x)
```

```python
# loop over a dictionary with keys() function
for x in carDict.keys():
  print(x)
```

```python
# loop over a dictionary to get the values one-by-one
for x in carDict:
  print(carDict[x])
```

```python
# loop over a dictionary with values() function
for x in carDict.values():
  print(x)
```

```python
# loop over a dictionary with items() function to get the key:value pair one-by-one
for x,y in carDict.items():
  print(f'{x} -> {y}')
```

## **Delete / Remove**

**using del directive and slicing operator**

```python
carDict =	{
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964,
  'color': ['red', 'green', 'white']
}

del carDict['brand']
print(carDict)
```

**Clear / Empty the dictionary**

```python
carDict =	{
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964,
  'color': ['red', 'green', 'white']
}

print(carDict)

carDict.clear()
print('After clearing the dictionary: ', carDict)
```

**Delete entire dictionary using del directive**

```python
carDict =	{
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964,
  'color': ['red', 'green', 'white']
}

print(carDict)

del carDict
print(f'after Deletion: {carDict}')    #this will cause an error because "thisdict" no longer exists.
```

**pop() method removes the item with the specified key name**

```python
carDict =	{
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964,
  'color': ['red', 'green', 'white']
}

print(carDict)

value = carDict.pop('year')
print(value)

print(f'After pop: {carDict}')
```

**popitem() method removes the last inserted item (in versions before 3.7, a random item is removed instead)**

```python
carDict =	{
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964,
  'color': ['red', 'green', 'white']
}

print(carDict)

value = carDict.popitem()
print(value)

print(f'After pop: {carDict}')
```
