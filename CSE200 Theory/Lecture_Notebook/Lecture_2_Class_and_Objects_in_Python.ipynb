```python
class Student(object):
  def __init__(self,name,id,section,cgpa):
    self.name = name
    self.id = id
    self.section = section
    self.cgpa = cgpa

student1 = Student('Rajit', 1234, 1, 2.99)
student2 = Student('Suha', 12345, 2, 3.98)

print(f'Name (before change): {student1.name}')
student1.name = 'Mahbubul Syeed'
print(f'Name (after change): {student1.name}')

print(f'Section (before change): {student2.section}')
student2.section = 4
print(f'Section (after change): {student2.section}')

# change the id of st1 to 9876
student1.id = 9876
print(student1.id)

# find the higher cgpa
if student1.cgpa  > student2.cgpa:
  print(f'{student1.name} has higher cgpa!')
else:
  print(f'{student2.name} has higher cgpa!')
```

```python
class Student(object):
  def __init__(self,name,id,section,cgpa):
    self.name = name
    self.id = id
    self.section = section
    self.cgpa = cgpa

  def printInfo(self):
    print(f'Name: {self.name}\nID:{self.id}\nSection:{self.section}\nCGPA:{self.cgpa}')

student1 = Student('Rajit', 1234, 1, 3.99)
student2 = Student('Suha', 12345, 2, 3.98)

student1.printInfo()
student2.printInfo()

hossain = Student('Hossain', 123, 1, 3.55)
hossain.printInfo()
```

```python
class Course(object):
  def __init__(self, courseid, coursename, credit):
    self.courseid = courseid
    self.coursename = coursename
    self.credit = credit

c1 = Course('csc101', 'Intro to Programming', 4)
print(c1.courseid, c1.coursename)

ds = Course('cse203', 'data structure', 4)
print(ds.courseid, ds.coursename)
ds.coursename = 'Data Structure'
print(ds.courseid, ds.coursename)
```

```python
class Course(object):
  def __init__(self, courseid, coursename, credit):
    self.courseid = courseid
    self.coursename = coursename
    self.credit = credit

  def setCourseID(self, newId):
    self.courseid = newId

  def getCoureID(self):
    return self.courseid

  def printAll(self):
    print(self.courseid, self.coursename, self.credit)

c1 = Course('csc101', 'Intro to Programming', 4)
ds = Course('cse203', 'data structure', 4)
c1.printAll()
ds.printAll()

ds.setCourseID('CSE203')
ds.printAll()
print(ds.getCoureID())
```
